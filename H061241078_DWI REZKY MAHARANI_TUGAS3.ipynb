### 1. Problem Statement
Data curah hujan stasiun sering memiliki nilai yang hilang (missing values) atau timestamp ganda. Diperlukan prosedur inspeksi non-destruktif untuk memeriksa struktur data tanpa mengubah file mentah.

### 2. Minimal Theory & Constraints
- Konvensi Waktu:Timestamp diperlakukan sebagai UTC.
- Invarian: Data mentah di `data/raw/` bersifat read-only dan tidak boleh diubah.

### 3. Ready-to-Use Prompt (P-IO-001)

```text
Bertindaklah sebagai Data Steward.

Tujuan:
Audit file CSV di [CSV_PATH] tanpa mengubah data mentah.

Batas & Konstran:
1. Baca kolom 'station_id', 'timestamp', dan 'rainfall_mm'.
2. Identifikasi jumlah baris, nilai yang hilang, dan timestamp duplikat.
3. Simpan laporan audit ke [REPORT_PATH].

````python
import pandas as pd
from pathlib import Path

# 1. Buat Data Sintetis
raw_data = """station_id,timestamp,rainfall_mm
ST-01,2026-01-01T00:00:00Z,12.5
ST-01,2026-01-02T00:00:00Z,0.0
ST-01,2026-01-02T00:00:00Z,0.0
ST-02,2026-01-01T00:00:00Z,
ST-02,2026-01-02T00:00:00Z,45.2"""

# 2. Simpan Data Mentah
raw_path = Path("data/raw/station_rainfall_synthetic.csv")
raw_path.parent.mkdir(parents=True, exist_ok=True)
raw_path.write_text(raw_data)

# 3. Proses Audit
df = pd.read_csv(raw_path)
missing_count = df['rainfall_mm'].isnull().sum()
duplicate_count = df.duplicated(subset=['station_id', 'timestamp']).sum()

# 4. Simpan Laporan Berkas
report_content = f"""# Audit Summary (R-IO-001)
- Total Baris: {len(df)}
- Missing Values: {missing_count}
- Duplicate Timestamps: {duplicate_count}
"""
report_path = Path("results/reports/io_audit_report.md")
report_path.parent.mkdir(parents=True, exist_ok=True)
report_path.write_text(report_content)

print("       OUTPUT DATA STASIUN CUACA          ")
print("\n1. Data Mentah (CSV):")
print(df.to_string(index=False))

print("\n2. Ringkasan Audit Kualitas Data:")
print(f" - Total Baris Data    : {len(df)}")
print(f" - Jumlah Data Kosong  : {missing_count}")
print(f" - Timestamp Duplikat : {duplicate_count}")
print("\n[Laporan berhasil disimpan ke: results/reports/io_audit_report.md]")
````
